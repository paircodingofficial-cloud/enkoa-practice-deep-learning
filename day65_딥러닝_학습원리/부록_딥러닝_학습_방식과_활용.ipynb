## 1. 인공지능, 머신러닝, 딥러닝, 생성형 AI의 관계

- **인공지능(AI, artificial intelligence):** 인식, 예측, 결정, 생성처럼 사람의 판단이 필요한 일을 컴퓨터가 하게 만든 프로그램 전체입니다. 사람이 규칙을 직접 적은 프로그램도 포함합니다.
- **머신러닝(machine learning):** 규칙을 사람이 적지 않고 데이터에서 학습하는 인공지능입니다.
- **딥러닝(deep learning):** 여러 층의 인공 신경망을 쓰는 머신러닝입니다(2절).
- **생성형 AI(generative AI):** 학습한 데이터와 비슷한 새 데이터(글, 그림, 소리, 코드)를 만드는 모델입니다. 오늘날 쓰이는 생성형 AI는 대부분 딥러닝 모델이라 그림에서 딥러닝 안에 둡니다.

2. **머신러닝:** 스팸과 정상으로 표시된 메일 수천 통에서 어떤 단어가 스팸과 함께 나오는지 학습합니다.

4. **생성형 AI:** 메일을 거르는 대신 받은 메일에 보낼 답장 초안을 새로 씁니다.

## 2. 딥러닝

### 2-1. 정의와 표현 학습

- **깊은(deep) 신경망:** 딥러닝은 여러 층의 인공 신경망(neural network)으로 데이터에서 직접 특징까지 학습합니다. 깊다는 층이 여러 겹이라는 뜻입니다.
- **특징(feature):** 예측에 쓰는 입력의 성질입니다. 타이타닉 과제의 나이, 객실 등급처럼 표의 열로 주어지기도 하고, 그림처럼 픽셀만 있으면 어떤 성질을 볼지 정해야 합니다.
- **표현 학습(representation learning):** 예측에 쓸 특징까지 모델이 학습하는 방식입니다. 전통적인 머신러닝은 사람이 입력에서 특징을 계산해 넣고, 딥러닝은 픽셀 같은 원래 입력을 넣어 층마다 특징을 학습합니다. 각 층의 출력을 그 입력의 표현(representation)이라고 부릅니다.

<img src="images/62_ml_vs_dl.png" width="1050" alt="제목은 '전통적인 머신러닝과 딥러닝: 특징을 정하는 주체가 다릅니다'입니다. 위아래 두 판이 같은 입력(손글씨 숫자 7 픽셀 그림)에서 시작해 같은 결과 7로 끝납니다. 위 전통적인 머신러닝 판에서는 입력이 '사람이 정한 특징' 카드(획의 수, 가로세로 비율, 구멍의 수)를 거쳐 분류 모델로 가고, 개발자가 특징 칩을 가리킵니다. 아래 딥러닝 판에서는 입력이 두 층의 신경망으로 바로 들어가고, '층이 학습한 특징' 상자 안 첫 층 아래에 짧은 선 조각과 '획', 둘째 층 아래에 가로 획 조각과 사선 획 조각, '모양'이 적혀 있습니다. 아래 띠에는 '딥러닝은 특징까지 데이터에서 학습합니다'가 적혀 있습니다.">

| 구분 | 필요한 데이터 양 | 잘 맞는 데이터 |
|---|---|---|
| 전통적인 머신러닝 | 비교적 적은 데이터로도 학습됩니다 | 열마다 뜻이 정해진 표 데이터(고객 정보, 거래 기록) |
| 딥러닝 | 파라미터가 많아 보통 더 많은 데이터가 필요합니다 | 그림, 글, 소리처럼 사람이 특징을 정하기 어려운 데이터 |

숫자 7 그림 한 장이 딥러닝 모델의 층을 지나는 순서로 따라가 봅니다. 층마다 무엇을 나타낼지는 학습으로 정해지며, 아래는 합성곱 신경망 같은 이미지 모델에서 흔히 보이는 경향입니다.

1. **입력(픽셀):** 28×28 그림의 784칸 밝기입니다. 밝기 값 하나만으로는 '7'이라는 정보가 드러나지 않습니다.
2. **앞쪽 층(획):** 짧은 선, 밝기가 바뀌는 경계처럼 그림의 작은 부분을 나타내는 특징을 계산합니다.
3. **뒤쪽 층(모양):** 앞 층의 특징을 조합해 윗부분 가로 획, 비스듬히 내려오는 긴 획 같은 더 큰 모양을 나타냅니다.
4. **출력층(숫자):** 열 칸의 로짓을 내고, 7 칸의 로짓이 가장 큽니다.

### 2-2. 학습 원리

딥러닝의 학습은 교안 01과 교안 02에서 한 반복 그대로입니다. 순전파로 예측하고, 손실을 계산하고, 역전파로 기울기를 구하고, 파라미터를 갱신합니다. 층이 많아도 `loss.backward()`가 모든 층의 기울기를 한 번에 구하므로 같은 학습 루프를 씁니다.

### 2-3. 널리 쓰이게 된 배경

- **데이터:** 인터넷과 디지털 기록으로 학습에 쓸 수 있는 그림, 글, 소리가 크게 늘었습니다.
- **GPU 계산:** 신경망 계산은 대부분 행렬 곱셈이라, GPU로 병렬 처리하면 CPU보다 훨씬 빨리 학습합니다.
- **학습 방법 개선:** 교안 02에서 쓴 ReLU 활성화 함수, Adam 옵티마이저처럼 층이 깊어도 학습이 잘 되게 하는 방법이 나왔습니다.
- **사례:** 2012년 발표된 AlexNet은 사진 약 130만 장을 1,000가지로 분류하는 신경망을 GPU로 학습해, 그때까지의 최고 성능보다 크게 앞선 결과를 냈습니다.

### 2-4. 대표 구조

세 구조 모두 여러 층의 신경망이고, 층 안의 계산 방식이 다릅니다.

- **DNN(deep neural network):** 한 층의 모든 뉴런이 다음 층의 모든 뉴런과 연결된 완전 연결 층(fully connected layer)을 쌓은 구조입니다. 교안 02의 784 → 128 → 64 → 10 모델이 DNN입니다.
- **CNN(convolutional neural network, 합성곱 신경망):** 합성곱 층을 쌓은 구조로, 이미지에 씁니다.
- **트랜스포머(Transformer):** 어텐션 층을 쌓은 구조로, 언어 처리와 LLM에 씁니다.

## 3. 지도학습

### 3-1. 정답이 있는 데이터

- **지도학습(supervised learning):** 입력과 정답의 짝으로, 입력을 넣으면 정답을 내도록 학습하는 방식입니다. 예측과 정답의 차이를 손실로 계산하고 그 손실을 줄입니다.
- **레이블(label):** 입력마다 붙은 정답입니다. 시험 점수, 숫자 그림의 실제 숫자, 승객의 생존 여부가 레이블입니다.

### 3-2. 분류와 회귀

- **분류(classification):** 정해진 범주 가운데 하나를 예측합니다. 출력층이 범주마다 로짓을 내고, softmax로 바꾼 확률이 가장 큰 범주를 답으로 고릅니다.
- **회귀(regression):** 연속된 숫자를 예측합니다. 출력은 숫자 하나이고 손실은 MSE나 MAE입니다.

교안 01의 회귀와 교안 02의 분류는 같은 네 단계를 거치고, 출력과 손실만 다릅니다.

| 4. 갱신 | 손실이 줄도록 `w`, `b`를 바꿈 | 손실이 줄도록 층마다 가중치를 바꿈 |

**과제의 학습 방식**
- **과제 01:** 타이타닉 승객을 사망과 생존 2개 범주로 나누는 분류입니다.
- **과제 02:** 날짜로 요일 7개 범주 가운데 하나를 맞히는 분류입니다.

## 4. 비지도학습

- **비지도학습(unsupervised learning):** 정답 없이 입력 데이터만으로 데이터 안의 구조를 찾는 방식입니다.
- **군집화(clustering):** 비슷한 데이터끼리 같은 군집(cluster)으로 묶습니다. 텍스트 임베딩 단원과 토픽 모델링 단원에서 문서 임베딩을 KMeans, HDBSCAN으로 묶은 것이 군집화입니다.
- **차원 축소(dimensionality reduction):** 열이 많은 데이터를 정보를 최대한 남기며 적은 열로 줄입니다. 텍스트 임베딩 단원과 토픽 모델링 단원에서 768차원 임베딩을 2차원으로 줄인 UMAP이 차원 축소입니다.

<img src="images/64_unsupervised_clustering.png" width="1050" alt="제목은 '비지도학습: 정답 없이 비슷한 데이터끼리 묶습니다'입니다. 두 판 모두 가로축이 구매 횟수, 세로축이 평균 구매 금액입니다. 왼쪽 군집 전 판에는 색이 없는 회색 점들이 세 덩어리로 흩어져 있고, 오른쪽 군집 후 판에는 같은 점들이 청록, 분홍, 갈색으로 칠해져 군집 1, 군집 2, 군집 3 점선 원으로 묶여 있습니다. 아래 띠에는 '군집의 뜻은 사람이 해석해 이름을 붙입니다'가 적혀 있습니다.">

쇼핑몰 고객을 구매 기록으로 묶는 예입니다.

1. **입력:** 고객마다 구매 횟수와 평균 구매 금액입니다. 고객 유형을 적은 정답 열은 없습니다.
2. **군집화:** 두 값을 표준화한 뒤 KMeans로 가까운 고객끼리 세 군집으로 나눕니다.
3. **해석:** 군집마다 평균값을 보고 사람이 이름(자주 사는 소액 고객, 가끔 사는 고액 고객)을 붙입니다.
4. **활용:** 군집마다 다른 쿠폰과 안내 메일을 보냅니다.

## 5. 강화학습

### 5-1. 구성 요소

- **강화학습(reinforcement learning):** 에이전트가 환경 속에서 행동하고 받은 보상으로, 보상의 합이 커지는 행동 방법을 학습하는 방식입니다. 정답 행동은 주어지지 않고 보상만 주어지며, 보상이 여러 행동 뒤에야 오기도 합니다.
- **에이전트(agent):** 행동을 고르는 학습 대상입니다.
- **환경(environment):** 에이전트가 행동하는 대상입니다. 게임, 바둑판, 시뮬레이터가 환경입니다.
- **상태(state):** 지금 환경이 어떤지 나타내는 정보입니다.
- **행동(action):** 에이전트가 한 상태에서 고를 수 있는 선택입니다.
- **보상(reward):** 행동 뒤에 환경이 주는 숫자입니다. 좋은 결과는 양수, 나쁜 결과는 음수로 줍니다.
- **정책(policy):** 상태를 받아 행동을 고르는 규칙입니다. 정책을 신경망으로 만드는 방법(알파고의 정책 네트워크)에서는 상태를 입력받아 행동마다 로짓을 내는 신경망이 정책입니다.

<img src="images/65_rl_loop.png" width="1050" alt="제목은 '강화학습: 행동하고 보상을 받으며 정책을 학습합니다'입니다. 왼쪽 주황 에이전트 카드 안에 회로 칩 모양 상자와 글자 '정책'이, 오른쪽 파랑 환경 카드 안에 4칸 x 4칸 격자 미로 게임이 있습니다. 미로의 왼쪽 아래 칸에 노란 공 모양 캐릭터, 오른쪽 위 칸에 초록 깃발, 지나갈 수 없는 회색 벽 칸 네 개가 있고 '출구 +10', '한 칸 이동 -1'이 적혀 있습니다. 에이전트에서 환경으로 주황 행동 화살표가, 환경에서 에이전트로 파랑 상태 화살표와 초록 보상 화살표가 돌아옵니다. 아래 띠에는 '받는 보상의 합이 커지도록 정책을 갱신합니다'가 적혀 있습니다.">

### 5-2. 미로 게임 예

1. **상태:** 캐릭터가 4×4 미로의 왼쪽 아래 칸에 있습니다.
2. **행동:** 위, 아래, 왼쪽, 오른쪽 가운데 하나로 움직입니다. 회색 벽 칸으로는 갈 수 없습니다.
3. **보상:** 한 칸 움직일 때마다 -1, 출구에 닿으면 +10을 받습니다. 그래서 돌아가는 길은 보상의 합이 작습니다.
4. **정책 갱신:** 시작부터 출구까지 한 번 진행한 것을 에피소드(episode)라고 합니다. 에피소드를 여러 번 진행하며 보상의 합이 컸던 에피소드에서 고른 행동의 확률을 높이면, 에피소드가 거듭될수록 출구까지 짧은 길을 고릅니다.

<img src="images/66_alphago_pluto.png" width="1050" alt="제목은 '강화학습 사례: 바둑과 실시간 전략 게임'입니다. 왼쪽 알파고 판에는 돌이 놓인 바둑판과 주황 원으로 표시한 다음 수 자리가 있고, '2016 이세돌 9단과 대국 4:1', '사람 기보 학습 뒤 자기 대국'이 적혀 있습니다. 오른쪽 플루토 판에는 위에서 내려다본 실시간 전략 게임 장면이 평면 그림으로 있습니다. 왼쪽 아래 파랑 진영과 오른쪽 위 빨강 진영의 건물과 동그라미 유닛, 가운데의 노랑 자원, 구석의 작은 미니맵이 보이고, '2026 스타크래프트: 브루드워 AI', '자기 대국 강화학습만으로 학습'이 적혀 있습니다.">

### 5-3. 알파고(AlphaGo)

- **결과:** 2016년 3월 서울에서 이세돌 9단과 다섯 판을 두어 4:1로 이겼습니다.
- **구조:** 다음 수를 고르는 정책 네트워크(policy network)와 판세를 평가하는 가치 네트워크(value network), 두 신경망을 씁니다.
- **학습:** 사람 고수의 기보로 다음 수를 예측하는 지도학습을 먼저 하고, 자기 자신과 대국(self-play)하며 강화학습을 했습니다. 대국 중에는 몬테카를로 트리 탐색(MCTS, Monte Carlo tree search)으로 두 신경망을 써서 수를 미리 따져 봅니다.
- **강화학습 구성:** 상태는 바둑판의 돌 배치, 행동은 돌을 놓을 자리, 보상은 대국의 승패입니다.

출처: [Google DeepMind, AlphaGo](https://deepmind.google/research/alphago/), [Silver 외 (2016), Mastering the game of Go with deep neural networks and tree search, Nature 529](https://doi.org/10.1038/nature16961)

### 5-4. 알파고 제로(AlphaGo Zero)

- **학습:** 사람 기보 없이 바둑 규칙만 주고, 자기 자신과 대국한 기록만으로 학습했습니다.
- **결과:** 자기 대국으로 3일 학습한 뒤, 이세돌 9단을 이긴 버전의 알파고를 100:0으로 이겼습니다(2017년 10월 발표).
- **의미:** 사람의 데이터 없이 보상(승패)만으로 사람 데이터로 시작한 모델보다 강해질 수 있음을 보였습니다.

출처: [Google DeepMind, AlphaGo Zero: Starting from scratch](https://deepmind.google/blog/alphago-zero-starting-from-scratch/), [Silver 외 (2017), Mastering the game of Go without human knowledge, Nature 550](https://doi.org/10.1038/nature24270)

### 5-5. 플루토(Pluto)

- **정의:** 실시간 전략 게임 스타크래프트: 브루드워(StarCraft: Brood War, 스타크래프트 1)를 플레이하는 AI입니다. 개발자 tscmoo가 GitHub에 실행 파일을 공개했고, CoG 2026 StarCraft AI Competition 출전작입니다.
- **학습:** 신경망 하나가 자기 대국 강화학습(self-play reinforcement learning)만으로 학습해, 운영(macro)과 전투 조작(micro)까지 경기 전체를 세 종족 모두로 플레이합니다. 사람이 정해 둔 스크립트가 없습니다.
- **강화학습 구성:** 상태는 경기 상황, 행동은 유닛과 건물에 내리는 명령이며, 모델은 6게임 프레임마다 한 번 행동을 고릅니다. 보상 설계는 공개 자료에 적혀 있지 않고, 자기 대국으로 이기는 쪽을 학습합니다.
- **모델:** 파라미터 3억 1,500만 개(int8)이며, GPU 없이 CPU로 추론합니다.
- **알파고 제로와의 연결:** 사람 기록 없이 자기 자신과 대국하며 학습하는 방식이 알파고 제로와 같습니다.
- **앞선 사례:** 알파스타(AlphaStar)는 2019년 스타크래프트 II에서 그랜드마스터 등급, 활동 플레이어의 99.8%보다 높은 순위에 올랐습니다. 사람 경기 기록으로 먼저 지도학습한 뒤 강화학습했다는 점이 자기 대국만 쓴 플루토와 다릅니다.

출처: [tscmoo, Pluto (GitHub)](https://github.com/tscmoo/pluto), [tscmoo, Pluto CoG 2026 entry (GitHub 릴리스)](https://github.com/tscmoo/pluto/releases/tag/cog2026-2578600), [Google DeepMind, AlphaStar: Grandmaster level in StarCraft II using multi-agent reinforcement learning](https://deepmind.google/blog/alphastar-grandmaster-level-in-starcraft-ii-using-multi-agent-reinforcement-learning/)

### 5-6. 로봇 제어

- **로봇 제어(robot control):** 걷기, 물건 다루기 같은 로봇 동작의 정책을 강화학습으로 학습합니다.
- **강화학습 구성:** 상태는 관절 각도와 속도, 센서 값, 행동은 관절 모터에 보내는 명령, 보상은 넘어지지 않고 목표 방향으로 이동한 정도입니다.
- **시뮬레이션에서 실제로(sim-to-real):** 실제 로봇으로 수백만 번 시도하기는 어려워, 시뮬레이션에서 학습한 정책을 실제 로봇에 옮깁니다.
- **도메인 무작위화(domain randomization):** 시뮬레이션과 현실의 차이를 줄이려고 마찰, 무게 같은 물리 값을 바꿔 가며 학습합니다.

1. **로봇 손과 루빅스 큐브(OpenAI, 2019):** 시뮬레이션에서만 학습한 모델로 실제 로봇 손이 루빅스 큐브를 조작했습니다. 환경을 점점 어렵게 무작위로 바꾸는 자동 도메인 무작위화(ADR, automatic domain randomization)를 썼습니다.
2. **사족 보행 로봇 ANYmal(ETH Zurich, 2020):** 시뮬레이션에서 강화학습한 보행 제어기를 추가 학습 없이 실제 로봇에 옮겨, 진흙, 눈, 돌무더기, 빽빽한 수풀 위를 걸었습니다. 제어기의 입력은 관절과 몸에서 재는 감각 신호(proprioception)입니다.
3. **두 발 로봇 축구(Google DeepMind, 2024):** 관절 20개인 휴머노이드 로봇이 시뮬레이션에서 강화학습한 뒤 추가 학습 없이 실제 로봇으로 1대1 축구를 했습니다. 스크립트로 만든 기준 동작보다 181% 빨리 걷고 302% 빨리 돌았습니다.

## 6. 생성형 AI

### 6-1. 새 데이터를 만드는 모델

- **텍스트:** LLM(large language model, 대규모 언어 모델)이 질문에 답하고 글을 요약, 번역합니다. ChatGPT 같은 대화 서비스가 LLM을 씁니다.
- **이미지:** 이미지 생성 모델이 문장 설명을 받아 그림을 만듭니다. 확산 모델(diffusion model)은 잡음에서 시작해 잡음을 조금씩 걷어 내며 그림을 만듭니다.
- **음성:** 음성 합성(TTS, text-to-speech) 모델이 글을 읽는 목소리를 만듭니다.
- **코드:** LLM이 설명을 받아 코드를 작성하고, 작성된 코드를 설명합니다.

<img src="images/67_genai_generate.png" width="1050" alt="제목은 '생성형 AI: 입력을 받아 새 데이터를 생성합니다'입니다. 입력, 생성 모델, 생성 결과 세 열에 텍스트, 이미지, 코드 세 행이 있습니다. 텍스트 행은 '회의록을 세 줄로 요약'이 LLM을 거쳐 세 줄 문서가, 이미지 행은 '바다 위 등대 그림'이 이미지 생성 모델을 거쳐 등대 그림이, 코드 행은 '목록을 정렬하는 함수'가 LLM을 거쳐 'def sort_items(items):' 코드가 됩니다. 오른쪽 아래에 개발자가 결과를 가리킵니다.">

LLM이 회의록을 요약하는 과정입니다.

1. **입력:** 회의록 원문과 지시 "세 줄로 요약"을 토큰(token)으로 나눠 넣습니다. 토큰은 모델이 글을 나누는 단위(단어나 단어 조각)입니다.
2. **생성:** 다음에 올 토큰을 하나 고르고, 고른 토큰을 입력 뒤에 붙여 다시 다음 토큰을 고릅니다.
3. **종료:** 끝을 뜻하는 토큰이 나오거나 정한 길이에 이르면 멈춥니다.
4. **확인:** 사람이 원문과 대조해 빠진 결정 사항이나 원문에 없는 내용(환각, hallucination)이 없는지 확인합니다.

### 6-2. LLM 학습 단계

LLM을 학습하는 대표 흐름으로, OpenAI의 InstructGPT 연구(2022)가 쓴 세 단계입니다.

1. **사전학습(pretraining):** 대량의 글에서 다음 토큰을 예측하도록 학습합니다. 정답(다음 토큰)을 글 자체에서 만들어 자기지도학습(self-supervised learning)이라고 부르며, 손실은 교안 02처럼 교차 엔트로피입니다.
2. **지도 파인튜닝(SFT, supervised fine-tuning):** 사람이 쓴 질문과 모범 답변의 짝으로 지도학습해, 지시를 따르는 답변을 내게 합니다.
3. **RLHF(reinforcement learning from human feedback):** 같은 질문의 답변들에 사람이 매긴 선호 순위를 보상으로 바꿔 LLM을 강화학습합니다.

<img src="images/68_llm_training_stages.png" width="1050" alt="제목은 'LLM 학습 단계: 사전학습, 지도 파인튜닝, RLHF'입니다. 카드 세 장이 화살표로 이어집니다. 1 사전학습(다음 토큰 예측) 카드에는 '오늘 날씨가 매우' 다음에 '맑다'가, 2 지도 파인튜닝(질문과 모범 답변) 카드에는 '휴가 신청 방법' 질문과 '모범 답변'이, 3 RLHF(사람의 선호 순위) 카드에는 답변 A와 B, 순위 'A > B', '보상 모델 점수'가 있습니다. 아래 띠에는 '1, 2는 정답과 비교하는 지도학습 방식, 3은 강화학습입니다'가 적혀 있습니다.">

- **사례:** InstructGPT 연구에서 사람 평가자는 이 세 단계로 학습한 13억 파라미터 모델의 답변을 1,750억 파라미터 GPT-3의 답변보다 더 선호했습니다.
- **다른 방법:** 선호를 반영하는 단계에는 강화학습 없이 선호 쌍으로 바로 학습하는 DPO 같은 방법도 씁니다. 이어지는 SFT 지도 파인튜닝 단원과 DPO 선호 정렬 단원이 이 단계들을 실습합니다.

출처: [Ouyang 외 (2022), Training language models to follow instructions with human feedback, arXiv:2203.02155](https://arxiv.org/abs/2203.02155)

## 7. 학습 방식과 생성형 AI 비교

| 구분 | 데이터 | 배우는 것 | 대표 활용 |
|---|---|---|---|
| 지도학습 | 입력과 정답의 짝 | 입력에서 정답으로 가는 대응 | 손글씨 숫자 분류, 시험 점수 예측 |
| 비지도학습 | 정답 없는 입력 | 데이터 안의 군집과 구조 | 고객 군집화, 문서 군집화 |
| 강화학습 | 행동의 결과로 받는 보상 | 보상의 합이 커지는 정책 | 알파고, 플루토, 로봇 제어, RLHF |
| 생성형 AI(만드는 결과로 붙인 이름) | 대량의 글과 그림, 모범 답변, 선호 순위 | 학습한 데이터와 비슷한 새 데이터를 만드는 법 | 요약, 그림 생성, 코드 작성 |

- **생성형 AI의 자리:** 위 세 줄은 학습 방식이고 생성형 AI는 만드는 결과로 붙인 이름입니다. LLM은 지도학습(자기지도학습 포함)과 강화학습을 단계마다 섞어 학습합니다(6-2).
- **고르는 기준:** 정답이 있으면 지도학습, 정답이 없고 군집이나 구조를 찾으려면 비지도학습, 정답 행동은 없고 결과의 좋고 나쁨만 숫자로 매길 수 있으면 강화학습, 결과물이 새 글이나 그림이면 생성형 AI를 씁니다.

## 8. 학습 방식별로 할 수 있는 일

**지도학습**
- **불량품 판별(분류):** 생산 라인의 부품 사진으로 정상과 불량을 나눠 검사를 자동화합니다.
- **부정 결제 탐지(분류):** 카드 결제 기록(금액, 시간, 가맹점)으로 부정 사용 여부를 판정합니다.
- **판매량 예측(회귀):** 지난 판매량, 요일, 날씨로 다음 주 판매량을 예측해 재고 주문량을 정합니다.
- **집값 예측(회귀):** 면적, 위치, 연식으로 거래 가격을 예측합니다.

**비지도학습**
- **상품 군집화:** 함께 팔리는 패턴이 비슷한 상품끼리 묶어 진열과 묶음 할인을 정합니다.
- **문의 주제 파악(군집화):** 고객 문의 글의 임베딩을 묶어 자주 나오는 주제를 찾고 FAQ를 만듭니다.
- **이상 탐지(anomaly detection):** 대부분의 기록과 멀리 떨어진 설비 센서 값이나 접속 기록을 찾아 점검 대상으로 올립니다.
- **데이터 시각화(차원 축소):** 열이 수백 개인 데이터를 2차원으로 줄여 그림으로 봅니다.

**강화학습**
- **게임 에이전트:** 바둑, 전략 게임, 게임 속 캐릭터의 행동을 학습합니다(알파고, 플루토).
- **로봇 제어:** 물류 창고 로봇의 물건 집기, 사족 로봇의 험지 보행을 시뮬레이션에서 학습해 실제 로봇에 옮깁니다(5-6).
- **추천 순서:** 사용자의 클릭과 시청 시간을 보상으로 추천 목록의 순서를 학습합니다.
- **LLM 답변 개선:** RLHF로 사람이 선호하는 답변 쪽으로 LLM을 학습합니다.

**생성형 AI**
- **문서 작업:** 회의록 요약, 보고서 초안, 번역을 만듭니다.
- **사내 질의응답:** 사내 문서를 검색해 근거와 함께 답합니다(RAG 단원의 검색 증강 생성).
- **개발 지원:** 코드와 테스트를 작성하고, 낯선 코드를 설명합니다.
- **이미지 제작:** 상품 이미지 시안, 교육 자료의 개념 그림을 만듭니다.

## 9. 핵심 정리

- **관계:** 인공지능 안에 머신러닝, 머신러닝 안에 딥러닝이 있고, 오늘날 생성형 AI는 대부분 딥러닝 모델입니다.
- **딥러닝:** 여러 층의 신경망이 특징을 직접 학습합니다(표현 학습). 학습은 교안 01, 02의 예측 → 손실 계산 → 역전파 → 갱신 반복입니다.
- **지도학습:** 입력과 정답의 짝으로 학습합니다. 분류는 범주(교차 엔트로피), 회귀는 숫자(MSE)를 예측하며, 이 단원의 교안과 과제는 모두 지도학습입니다.
- **비지도학습:** 정답 없이 데이터 안의 구조를 찾습니다. 군집화와 차원 축소가 대표입니다.
- **강화학습:** 에이전트가 상태를 보고 행동해 받은 보상으로 정책을 학습합니다. 알파고는 이세돌 9단과의 대국에서 4:1로 이겼고, 알파고 제로와 플루토(스타크래프트: 브루드워)는 사람 기록 없이 자기 대국만으로 학습했습니다. 알파고 제로는 그 알파고를 100:0으로 이겼습니다. 로봇 제어는 시뮬레이션에서 학습한 정책을 실제 로봇에 옮기며(sim-to-real), 도메인 무작위화로 현실과의 차이를 줄입니다.
- **생성형 AI:** 학습한 데이터와 비슷한 새 데이터를 만듭니다. LLM 학습의 대표 흐름은 InstructGPT의 사전학습(다음 토큰 예측), 지도 파인튜닝, RLHF입니다.
